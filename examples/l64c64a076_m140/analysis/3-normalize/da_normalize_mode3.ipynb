## User Inputs

These fields describe a downstream normalization job that consumes grouped DA fit/sample outputs from the N-state fit workflow.


```python
DA_FIT_RESULTS = REPO_ROOT / "examples" / "l64c64a076_m140" / "analysis" / "2-bm" / "results_fit_nst2"

workflow_config = {
    # Data / metadata settings
    "title_pattern": "l64c64a076_m140_fit_pz*",
    "input_root": str(DA_FIT_RESULTS),
    "ns": 64,
    "lattice_spacing_fm": 0.076,
    "pzlist": [0, 1, 5, 6, 7, 8],
    "gmlist": ["T5"],
    "etalist": ["eta0"],
    "bTlist": [0],
    "bzlist": [bz for bz in range(0, 21)],
    "component": "real",
    # Run again with component = "imag" to produce the imaginary part.
    "nstates": 2,

    # Normalization settings
    "normalization_mode": "mode3",

    # Output settings
    "plot": True,
    "results_dir": "/Users/xiang/Desktop/codes/lat-hadron-analysis-DA/examples/l64c64a076_m140/analysis/3-normalize",
}
```

## Option Guide

- `input_root`: Root directory containing existing grouped DA fit/sample outputs.
- `title_pattern`: Same pattern used by the DA fit workflow, so the normalization step can find per-`pz` dataset directories.
- `component`, `nstates`: Select which grouped fit/sample outputs to normalize.
- `normalization_mode`: One of `"mode1"`, `"mode2"`, or `"mode3"`.
- `mode1`: `m0(pz, bT, bz) / m0(pz, bT=0, bz=0)`. The reference point is taken at the same `pz`, with `bT=0` and `bz=0`.
- `mode2`: `m0(pz, bT, bz) / m0(pz=0, bT, bz=0)`. The reference point is taken at `pz=0`, with the same `bT`, and `bz=0`.
- `mode3`: `(m0(pz, bT, bz) / m0(pz, bT=0, bz=0)) / (m0(pz=0, bT, bz=0) / m0(pz=0, bT=0, bz=0))`. This combines the `mode1` and `mode2` references into a double ratio.
- `bTlist` / `bzlist`: Which grouped matrix-element entries to normalize and rewrite.
- `plot`: Optional convenience plot of normalized `m0` vs `bz`.
