## Imports / Setup


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
SRC_DIR = REPO_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from lqcd_analysis.notebook_workflows import (
    pretty_print_config,
    run_plot_2pt_from_notebook,
    validate_plot_2pt_notebook_config,
)
```

## User Inputs

This template uses the same single-config notebook pattern as the analysis templates.
These defaults assume you already ran the N-state example and generated outputs under `examples/outputs/nstate_fit_realdata/`.


```python
NSTATE_RESULTS = REPO_ROOT / "examples" / "outputs" / "nstate_fit_realdata" / "l64c64a076_m140_fit_k0_pz0"

workflow_config = {
    "output_dir": str(REPO_ROOT / "examples" / "outputs" / "plot_2pt_notebook"),
    "correlator_table": str(NSTATE_RESULTS / "tables" / "l64c64a076_m140_fit_k0_pz0_normal_correlator_mean.txt"),
    "meff_table": str(NSTATE_RESULTS / "tables" / "l64c64a076_m140_fit_k0_pz0_normal_effective_mass_tmax12.txt"),
    "fit_table": str(NSTATE_RESULTS / "tables" / "l64c64a076_m140_fit_k0_pz0_normal_2state_tmax12_fits.txt"),
    "nstates": 2,
    "model": "normal",
    "title": "l64c64a076_m140_fit_k0_pz0",
    "nt": 64,
    "lattice_spacing_fm": 0.076,
}
```

## Option Guide

Edit only `workflow_config` in the cell above for normal usage.

- `output_dir`: Directory where figures and plot products will be written.
- `correlator_table`: Path to the mean correlator table produced by the N-state fit workflow.
- `meff_table`: Path to the effective-mass table produced by the N-state fit workflow.
- `fit_table`: Path to a specific fit table, for example a `1state` or `2state` result.
- `nstates`: Number of states represented by `fit_table`.
  Choices: `1`, `2`, or `3`, and it should match the selected fit table.
- `model`: Correlator model label used in titles and plot interpretation.
  Choices normally match the fit workflow: `"normal"`, `"symmetric"`, or `"antisymmetric"`.
- `title`: Dataset title shown in figure titles and output names.
- `nt`: Temporal extent `Nt`, needed for model reconstruction plots.
- `lattice_spacing_fm`: Lattice spacing in fm, used to convert plotted energies from `E*a` to physical units in MeV.

Practical note:
- When switching between 1-state, 2-state, and 3-state plots, update both `fit_table` and `nstates` together.


## Input Summary / Validation

The config below is the one place you usually need to edit.
Validate it before running to catch missing paths or mismatched plotting metadata early.


```python
print(pretty_print_config(workflow_config))
validated_plot_config = validate_plot_2pt_notebook_config(workflow_config)
validated_plot_config
```

## Run Analysis


```python
plot_outputs = run_plot_2pt_from_notebook(workflow_config)
for path in plot_outputs:
    print(path)
```
