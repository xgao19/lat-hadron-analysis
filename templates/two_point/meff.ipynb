# Two-Point Effective Mass Template

Standalone effective-mass workflow for two-point correlators.
It reads correlator CSV files, applies folding, bootstraps the data, and writes effective-mass tables.


## Imports / Setup


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
SRC_DIR = REPO_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

import matplotlib
matplotlib.rcParams.update({
    "font.family": "Times New Roman",
    "mathtext.fontset": "custom",
    "mathtext.rm": "Times New Roman",
    "mathtext.it": "Times New Roman:italic",
    "mathtext.bf": "Times New Roman:bold",
})

from lqcd_analysis.notebook_workflows import (
    pretty_print_config,
    render_effective_mass_input_text,
    run_effective_mass_from_notebook,
    validate_effective_mass_notebook_config,
)
```

## User Inputs


```python
EXAMPLE_DATA = REPO_ROOT / 'examples' / 'data' / 'l64c64a076_m140' / 'c2pt_csv'
EXAMPLE_OUTPUTS = REPO_ROOT / 'examples' / 'outputs' / 'effective_mass_notebook'

workflow_config = {
    'title_pattern': 'l64c64a076_m140_fit_k0_pz*',
    'ns': 64,
    'nt': 64,
    'lattice_spacing_fm': 0.076,
    'c2pt': str(EXAMPLE_DATA / 'c2pt_5_5_k0_pz*_real.csv'),
    'pzlist': [0],
    'fold_t': 'periodic',
    'tsrange': [0, 24],
    'model': 'symmetric',
    'binsize': 1,
    'bootstrap_samples': 64,
    'bootstrap_size': 64,
    'seed': 2026,
    'results_dir': str(EXAMPLE_OUTPUTS),
}
```

## Option Guide

- `title_pattern`: Output title pattern. Use `*` where the momentum index should be inserted.
- `c2pt`: Correlator CSV path or wildcard pattern.
- `pzlist`: List of momenta to analyze.
- `fold_t`: Folding mode before bootstrapping.
- `tsrange`: Time range to retain after folding.
- `model`: `normal`, `symmetric`, or `antisymmetric`.
- `binsize`, `bootstrap_samples`, `bootstrap_size`, `seed`: Bootstrap controls.
- `results_dir`: Optional output directory override.


```python
validated = validate_effective_mass_notebook_config(workflow_config)
print(pretty_print_config(validated))
print(render_effective_mass_input_text(validated))
```

## Run Workflow


```python
outputs = run_effective_mass_from_notebook(validated)
outputs
```
