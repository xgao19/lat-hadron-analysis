## User Inputs

These fields describe a downstream Fourier-transform job that consumes grouped TMDWF fit/sample outputs.
Run the TMDWF N-state fit first so the grouped `fit_table` and `sample_table` already exist.


```python
TMDWF_RESULTS = REPO_ROOT / "examples" / "outputs" / "tmdwf_fit_notebook" / "l64c64a076_m140_tmdwf_pz0"
FOURIER_OUTPUTS = REPO_ROOT / "examples" / "outputs" / "tmdwf_fourier_notebook"

workflow_config = {
    # Data / metadata settings
    "title": "l64c64a076_m140_tmdwf_pz0_T5_eta0",
    "stem": "l64c64a076_m140_tmdwf_pz0_T5_eta0_bT0",
    "pz": 0,
    "ns": 64,
    "lattice_spacing_fm": 0.076,
    "component": "real",
    "nstates": 1,
    "bT": 0,
    "fit_table": str(TMDWF_RESULTS / "tables" / "l64c64a076_m140_tmdwf_pz0_T5_eta0_bT0_real_1state_fit.txt"),
    "sample_table": str(TMDWF_RESULTS / "samples" / "l64c64a076_m140_tmdwf_pz0_T5_eta0_bT0_real_1state_samples.txt"),

    # Fourier-transform parameter settings
    "x_range": [-0.5, 1.5],
    "x_count": 201,
    "zstep_fm": 0.01,
    "interpolation_kind": "cubic",

    # Output settings
    "plot": True,
    "results_dir": str(FOURIER_OUTPUTS),
}
```

## Option Guide

Edit only `workflow_config` in the cell above for normal usage.

- `title`, `stem`: Labels for the downstream Fourier job. `stem` controls the output filename prefix.
- `fit_table`, `sample_table`: Existing grouped TMDWF outputs from the N-state fit workflow. The Fourier step reads `m0_mean`, `m0_err`, and grouped bootstrap `m0` samples from these files.
- `component`: Choose `"real"` or `"imag"`.
- `nstates`, `bT`, `pz`, `ns`, `lattice_spacing_fm`: Physics/input metadata needed by the cosine transform.
- `x_range`, `x_count`: Convenient way to build a regular `x` grid. If you want a custom irregular grid, replace these with `x_values`.
- `zstep_fm`: Refined integration step in fm. The default downstream workflow uses `0.01` fm.
- `interpolation_kind`: Interpolation used before integrating on the refined `z` grid. Default is `"cubic"`.
- `plot`: Whether to also write the Fourier PDF plot.
- `results_dir`: Output directory for Fourier tables, sample files, and plots.
