# TMDWF Normalize Template

This notebook is a standalone wrapper around the repository's post-fit TMDWF normalization workflow.
It reads grouped TMDWF fit/sample outputs, computes normalized matrix elements sample-by-sample, and writes normalized grouped outputs that can feed the downstream Fourier workflow.


## Imports / Setup


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
SRC_DIR = REPO_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from lqcd_analysis.notebook_workflows import (
    pretty_print_config,
    render_tmdwf_normalize_input_text,
    run_tmdwf_normalize_from_notebook,
    validate_tmdwf_normalize_notebook_config,
)
```

## User Inputs

These fields describe a downstream normalization job that consumes grouped TMDWF fit/sample outputs from the N-state fit workflow.


```python
EXAMPLE_TMDWF_FIT_RESULTS = REPO_ROOT / "examples" / "outputs" / "tmdwf_fit_notebook"
EXAMPLE_OUTPUTS = REPO_ROOT / "examples" / "outputs" / "tmdwf_normalize_notebook"

workflow_config = {
    # Data settings
    "title_pattern": "l64c64a076_m140_fit_pz*",
    "input_root": str(EXAMPLE_TMDWF_FIT_RESULTS),
    "ns": 64,
    "lattice_spacing_fm": 0.076,
    "pzlist": [0],
    "gmlist": ["T5"],
    "etalist": ["eta0"],
    "bTlist": [0],
    "bzlist": [0],
    "component": "real",
    "nstates": 2,

    # Normalization settings
    "normalization_mode": "mode1",

    # Output settings
    "plot": False,
    "results_dir": str(EXAMPLE_OUTPUTS),
}
```

## Option Guide

- `input_root`: Root directory containing existing grouped TMDWF fit/sample outputs.
- `title_pattern`: Same pattern used by the TMDWF fit workflow, so the normalization step can find per-`pz` dataset directories.
- `component`, `nstates`: Select which grouped fit/sample outputs to normalize.
- `normalization_mode`: One of `"mode1"`, `"mode2"`, or `"mode3"`.
- `mode1`: `m0(pz, bT, bz) / m0(pz, bT=0, bz=0)`. The reference point is taken at the same `pz`, with `bT=0` and `bz=0`.
- `mode2`: `m0(pz, bT, bz) / m0(pz=0, bT, bz=0)`. The reference point is taken at `pz=0`, with the same `bT`, and `bz=0`.
- `mode3`: `(m0(pz, bT, bz) / m0(pz, bT=0, bz=0)) / (m0(pz=0, bT, bz=0) / m0(pz=0, bT=0, bz=0))`. This combines the `mode1` and `mode2` references into a double ratio.
- `bTlist` / `bzlist`: Which grouped matrix-element entries to normalize and rewrite.
- `plot`: Optional convenience plot of normalized `m0` vs `bz`.
